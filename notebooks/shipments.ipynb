# Shipments

Not all deliveries originate at, or need to return to, a depot.
Sometimes a pickup needs to happen at one location, and be delivered to another, without any depot involvement.
We call this routing problem *pickup and delivery*, and model it using shipments.

```python
import pyvrp
import pyvrp.plotting
import pyvrp.stop
```

TODO

```python
COORDS = [(456, 320), (312, 418), (114, 80), (570, 160), (684, 240)]
DEMANDS = [5, 8, 7, 6]
```

```python
m = pyvrp.Model()
m.add_depot(m.add_location(*COORDS[0]))
m.add_vehicle_type(1, capacity=15)

for idx in [1, 3]:
    m.add_shipment(
        pickup_location=m.add_location(*COORDS[idx]),
        delivery_location=m.add_location(*COORDS[idx + 1]),
        amount=DEMANDS[idx - 1],
    )

for frm in m.locations:
    for to in m.locations:
        distance = abs(frm.x - to.x) + abs(frm.y - to.y)
        m.add_edge(frm, to, distance=distance)

res = m.solve(stop=pyvrp.stop.MaxRuntime(1))
```

```python
print(res.best)
```

```python
pyvrp.plotting.plot_solution(res.best, m.data(), plot_locations=True)
```
